# 02. CIFAR-100 Exploratory Data Analysis & Integrity

This notebook downloads CIFAR-100, validates class balance across 100 fine labels, checks for corrupted/NaN samples, inspects spatial dimensions, and previews sample distributions.

```python
from cv_benchmark.data.dataset import CIFAR100DatasetWrapper, CIFAR100_FINE_CLASSES

wrapper = CIFAR100DatasetWrapper(data_dir="./data")
summary = wrapper.download_and_verify()
for k, v in summary.items():
    if k != "classes":
        print(f"{k}: {v}")
```

```python
import matplotlib.pyplot as plt
from torchvision.datasets import CIFAR100

dataset = CIFAR100(root="./data", train=True, download=False)
fig, axes = plt.subplots(4, 8, figsize=(16, 8))
for i, ax in enumerate(axes.flat):
    img, target = dataset[i]
    ax.imshow(img)
    ax.set_title(CIFAR100_FINE_CLASSES[target], fontsize=8)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
